### The RK4 Algorithm (Vectorized):
For a system $\mathbf{w}' = \mathbf{f}(t, \mathbf{w})$, where $\mathbf{w}$ is a vector, the step update is:
$$\mathbf{k}_1 = h \cdot \mathbf{f}(t_n, \mathbf{w}_n)$$
$$\mathbf{k}_2 = h \cdot \mathbf{f}(t_n + \frac{h}{2}, \mathbf{w}_n + \frac{\mathbf{k}_1}{2})$$
$$\mathbf{k}_3 = h \cdot \mathbf{f}(t_n + \frac{h}{2}, \mathbf{w}_n + \frac{\mathbf{k}_2}{2})$$
$$\mathbf{k}_4 = h \cdot \mathbf{f}(t_n + h, \mathbf{w}_n + \mathbf{k}_3)$$
$$\mathbf{w}_{n+1} = \mathbf{w}_n + \frac{1}{6}(\mathbf{k}_1 + 2\mathbf{k}_2 + 2\mathbf{k}_3 + \mathbf{k}_4)$$
